```python
import itertools
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

# -----------------------------
# 1️⃣ Define a simple 3-node MRF
# -----------------------------
G = nx.Graph()
G.add_edges_from([("A", "B"), ("B", "C")])

# visualize the network
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels=True, node_color='skyblue', node_size=2000, font_size=14, font_weight='bold')
plt.title("Simple Markov Random Field (A–B–C)")
plt.show()

# ---------------------------------
# 2️⃣ Define clique potentials φ(A,B) and φ(B,C)
# ---------------------------------
# Assume binary nodes: 0 or 1
phi_AB = np.array([[2.0, 1.0],   # φ(A,B)
                   [1.0, 3.0]])

phi_BC = np.array([[3.0, 1.0],   # φ(B,C)
                   [1.0, 2.0]])

# ---------------------------------
# 3️⃣ Compute partition function Z
# ---------------------------------
Z = 0
states = list(itertools.product([0, 1], repeat=3))  # all possible (A,B,C)

for A, B, C in states:
    potential = phi_AB[A, B] * phi_BC[B, C]
    Z += potential

print(f"Partition function Z = {Z:.4f}")

# ---------------------------------
# 4️⃣ Compute normalized probabilities P(A,B,C)
# ---------------------------------
probs = {}
for A, B, C in states:
    phi_val = phi_AB[A, B] * phi_BC[B, C]
    P = phi_val / Z
    probs[(A,B,C)] = P

print("\nJoint probabilities P(A,B,C):")
for k,v in probs.items():
    print(f"{k}: {v:.4f}")

# ---------------------------------
# 5️⃣ Example: marginal probability of A=1
# ---------------------------------
P_A1 = sum(p for (A,B,C), p in probs.items() if A == 1)
print(f"\nMarginal P(A=1) = {P_A1:.4f}")
```
