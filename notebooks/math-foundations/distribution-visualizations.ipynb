#chi-square

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# List of degrees of freedom to plot
k_values = [2, 5, 10, 20]

# Set up the figure
plt.figure(figsize=(10, 6))

# Generate and plot histograms for each k
for k in k_values:
    samples = np.random.chisquare(k, size=10000)  # Generate random chi-square values
    plt.hist(samples, bins=50, density=True, alpha=0.4, label=f'Histogram (k={k})')

    # Plot theoretical chi-square PDF
    x = np.linspace(0, max(samples), 1000)
    plt.plot(x, stats.chi2.pdf(x, k), lw=2, label=f'Theoretical $\chi^2_{k}$')

# Labels and title
plt.xlabel('x')
plt.ylabel('Density')
plt.title('Chi-Square Distributions for Different Degrees of Freedom')
plt.legend()
plt.grid(True)
plt.show()
```

#t- distribution

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Define x values
x = np.linspace(-4, 4, 1000)

# Standard normal (Z) distribution
z_pdf = stats.norm.pdf(x)

# Different degrees of freedom (df) for t-distribution
dfs = [1,5,9, 100]  # Higher df to see better convergence

plt.figure(figsize=(8, 5))

# Plot t-distributions for different degrees of freedom
for df in dfs:
    t_pdf = stats.t.pdf(x, df)
    plt.plot(x, t_pdf, label=f't-distribution (df={df})')

# Plot standard normal distribution
plt.plot(x, z_pdf, 'k--', label="Standard Normal (Z)", linewidth=2)

plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("t-Distribution Approaching Normal Distribution")
plt.legend()
plt.grid()
plt.show()
```
