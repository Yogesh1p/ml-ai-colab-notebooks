```python
# prompt: do mle estimation and optimize the parameters taking some sample datapoints

import numpy as np
from scipy.optimize import minimize

def log_likelihood(params, data):
    """
    Calculates the log-likelihood of the data given the parameters.

    Args:
        params: A NumPy array of parameters.
        data: A NumPy array of data points.

    Returns:
        The negative log-likelihood. (Minimization problem).
    """
    mu, sigma = params
    n = len(data)
    log_like = -0.5 * n * np.log(2 * np.pi * sigma**2) - 0.5 * np.sum(((data - mu) / sigma)**2)
    return -log_like  # Return the negative for minimization


# Sample data points
data = np.array([1.2, 2.5, 1.8, 3.1, 2.0])

# Initial guess for parameters (mu, sigma)
initial_guess = np.array([2.0, 1.0])

# Perform maximum likelihood estimation using the minimize function
result = minimize(log_likelihood, initial_guess, args=(data,))

# Print the optimized parameters (MLE estimates)
print("MLE estimates:")
print("mu:", result.x[0])
print("sigma:", result.x[1])
```
