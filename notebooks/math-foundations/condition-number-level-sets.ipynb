```python
import numpy as np
import matplotlib.pyplot as plt

def plot_quadratic_level_sets(A, title=""):
    # Create a grid of (x1, x2) points
    x1 = np.linspace(-3, 3, 300)
    x2 = np.linspace(-3, 3, 300)
    X1, X2 = np.meshgrid(x1, x2)
    X = np.stack([X1, X2], axis=-1)

    # Compute f(x) = x^T A x for each point
    Z = np.einsum('...i,ij,...j->...', X, A, X)

    # Plot
    plt.contour(X1, X2, Z, levels=20, cmap='viridis')
    plt.title(title)
    plt.xlabel("x₁")
    plt.ylabel("x₂")
    plt.axis('equal')
    plt.grid(True)

# Define two matrices: one well-conditioned (κ=1), one ill-conditioned (κ>>1)
A1 = np.array([[1, 0],
               [0, 1]])  # κ = 1, circular level sets

A2 = np.array([[10, 0],
               [0, 1]])  # κ = 10, elongated ellipse

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plot_quadratic_level_sets(A1, title="Well-Conditioned (κ = 1)")

plt.subplot(1, 2, 2)
plot_quadratic_level_sets(A2, title="Ill-Conditioned (κ = 10)")

plt.tight_layout()
plt.show()
```
