```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.random import default_rng

def plot_multivariate_normal(mu, cov, num_points=1000):
    """
    Generate and plot points from a multivariate normal distribution.

    Parameters:
    - mu: Mean vector (1D array)
    - cov: Covariance matrix (2D array)
    - num_points: Number of points to generate
    """
    # Generate random points from the multivariate normal distribution
    rng = default_rng()
    points = rng.multivariate_normal(mu, cov, num_points)

    # Plot the generated points
    plt.scatter(points[:, 0], points[:, 1], alpha=0.5)
    plt.title(f'Multivariate Normal Distribution\nMean: {mu}\nCovariance Matrix: {cov}')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.grid(True)
    plt.show()

# Example parameters
mu = np.array([0, 0])  # Mean vector
cov = np.array([[1, 0.8], [0.8, 1]])  # Covariance matrix

# Plot points
plot_multivariate_normal(mu, cov)
```
